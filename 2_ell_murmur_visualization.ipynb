```python
from matplotlib import pyplot as plt
import os 
import pickle 
import numpy as np
import pandas as pd
```

```python
pklfiles = sorted(os.listdir('processed_data'))
pklfiles 
```

```python
dfs = [] 
for pklfile in [file for file in pklfiles if 'df' in file]:
    with open(f'processed_data/{pklfile}', 'rb') as f:
        df = pickle.load(f)
        dfs.append(df)
len(dfs)
```

```python
df = pd.concat(dfs).reset_index(drop=True)
df 
```

```python
# if df is okay, delete the pkl files to save memory 
del(dfs)
```

```python
# with open('df_1_primeindex5000.pkl', 'rb') as f:
#     df = pickle.load(f)

with open('processed_data/5000_primes.pkl', 'rb') as f:
    primes = pickle.load(f)
```

```python
columns_ap = [f'a{p}' for p in primes]
df_filtered_for_aps = df.filter(items = columns_ap)
```

```python
# Modify n_min and n_max to get a plot
#   for the conductor window [n_min, n_max]
#############################
n_min = 30000
n_max = 100000
#############################

df_filtered_for_aps_and_conductors_0 = df_filtered_for_aps[(df['conductor'] >= n_min) & (df['conductor'] <= n_max) & (df['rank'] == 0)]
df_filtered_for_aps_and_conductors_1 = df_filtered_for_aps[(df['conductor'] >= n_min) & (df['conductor'] <= n_max) & (df['rank'] == 1)]

values_aps_for_conductor_window_0 = np.average(df_filtered_for_aps_and_conductors_0.values, axis=0)
values_aps_for_conductor_window_1 = np.average(df_filtered_for_aps_and_conductors_1.values, axis=0)

yaxisrange = max(abs(values_aps_for_conductor_window_0).max(), abs(values_aps_for_conductor_window_1).max())

plt.figure(figsize=(20,8))
plt.scatter(range(len(primes)),
            values_aps_for_conductor_window_0,
            label='rank 0', s =5)
plt.scatter(range(len(primes)),
            values_aps_for_conductor_window_1,
            label='rank 1', s=5)
plt.legend()
plt.ylim(-yaxisrange, yaxisrange)

plt.title(f'1st ~ 1000th primes, conductor in [{n_min}, {n_max}]')
plt.xlabel('index of prime p')
plt.ylabel('average of a_p')
plt.show()
```
